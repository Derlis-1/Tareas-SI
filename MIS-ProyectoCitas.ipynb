```python
import pandas as pd


# Extracción de datos

pacientes_cols = ['cedula', 'nombre_completo', 'edad', 'telefono', 'contrasena']
medicos_cols = ['codigo', 'nombre', 'edad_medico', 'especialidad', 'horario', 'contrasena_med']
citas_cols = ['id', 'id_paciente', 'codigo_medico', 'fecha', 'motivo', 'estado']

df_pacientes = pd.read_csv('data_pacientes.txt', sep='|', names=pacientes_cols, dtype={'cedula': str, 'telefono': str})
df_medicos = pd.read_csv('data_medicos.txt', sep='|', names=medicos_cols)
df_citas = pd.read_csv('data_citas.txt', sep='|', names=citas_cols, dtype={'id_paciente': str})


# 2. transformación e integración

df_analisis = df_citas.merge(df_pacientes, left_on='id_paciente', right_on='cedula', how='left')
df_analisis = df_analisis.merge(df_medicos, left_on='codigo_medico', right_on='codigo', how='left')


# 3. Generación de reportes en tablas


print("*******************************************************")
print("     MIS - REPORTES GERENCIALES ")
print("*******************************************************\n")

# TABLA A: Tasa de Cancelación / Inasistencia
print("[A] TASA DE EFECTIVIDAD DE CITAS:")
# Mapeo de los estados y convertir el conteo en una tabla
tabla_estados = df_citas['estado'].map({0: 'Efectiva', 1: 'Cancelada'}).value_counts().reset_index()
tabla_estados.columns = ['Estado de la Cita', 'Cantidad']
# Agregamos una columna extra de porcentaje para darle más valor gerencial
tabla_estados['Porcentaje'] = (tabla_estados['Cantidad'] / len(df_citas) * 100).round(1).astype(str) + '%'

# Imprimimos la tabla sin los índices numéricos laterales
print(tabla_estados.to_string(index=False))
print("-" * 50)


#TABLA B: Especialidades más Demandadas
print("\n[B] DEMANDA POR ESPECIALIDAD:")
tabla_especialidades = df_analisis['especialidad'].value_counts().reset_index()
tabla_especialidades.columns = ['Especialidad Médica', 'Total de Citas']

print(tabla_especialidades.to_string(index=False))
print("-" * 50)


#TABLA C: Distribución de Edades de Pacientes Atendidos
print("\n[C] DISTRIBUCIÓN ETARIA DE PACIENTES ATENDIDOS:")
bins = [0, 17, 64, 120]
etiquetas = ['Pediátricos [0-17]', 'Adultos [18-64]', 'Geriátricos [65+]']
df_analisis['grupo_etario'] = pd.cut(df_analisis['edad'], bins=bins, labels=etiquetas)

tabla_edades = df_analisis['grupo_etario'].value_counts().reset_index()
tabla_edades.columns = ['Grupo Etario', 'Pacientes Agendados']

print(tabla_edades.to_string(index=False))
print("*******************************************************")
```
